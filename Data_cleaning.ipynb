# Text Preproccesing

## Spliting the 'genres' and 'team' column

```python
df1['Genres'] = df1['Genres'].str.replace('[\[\]\']', '', regex=True).str.split(',')
df1 = df1.explode('Genres')
```

```python
df1['Team'] = df1['Team'].str.replace('[\[\]\']', '', regex=True).str.split(',')
df1 = df1.explode('Team')
```

## Converting the 'K' for the values in the rows for a particular column by multiplying by 1000

```python
cols = ['Times Listed', 'Number of Reviews', 'Plays', 'Playing', 'Backlogs', 'Wishlist']
for col in cols:
  if df1[col].astype(str).str.contains('K').any():
    df1[col] = df1[col].astype(str).str.replace('K', '', regex=False)
    df1[col] = pd.to_numeric(df1[col]) * 1000

display(df1.head())

for col in cols:
  df1[col] = df1[col].astype(int)
```

## Convert the release date into datetime value

```python
df1['Release Date'] = pd.to_datetime(df1['Release Date'], errors='coerce')
```
